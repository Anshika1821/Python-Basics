# Theoretical Answers
## Question 1: What is Python, and why is it popular?
Python is a high-level, interpreted programming language known for its simplicity and readability. It is popular because of its:
- Versatile applications (web development, data analysis, AI, etc.)
- Extensive libraries and frameworks
- Ease of learning and use
- Community support

## Question 2: What is an interpreter in Python?
An interpreter in Python is a program that executes Python code line by line, converting it into machine code at runtime.

## Question 3: What are pre-defined keywords in Python?
Pre-defined keywords are reserved words in Python that have specific meanings and purposes, such as `if`, `else`, `for`, `while`, `class`, etc.

## Question 4: Can keywords be used as variable names?
No, keywords cannot be used as variable names because they are reserved for specific purposes in Python.

## Question 5: What is mutability in Python?
Mutability refers to the ability of an object to be changed after it is created. Mutable objects (e.g., lists) can be modified, whereas immutable objects (e.g., tuples) cannot.

## Question 6: Why are lists mutable, but tuples are immutable?
Lists are mutable to allow flexibility in modifying elements, while tuples are immutable to ensure hashability and consistency when used as dictionary keys or in sets.

## Question 7: What is the difference between `==` and `is` operators in Python?
- `==`: Checks for value equality.
- `is`: Checks for object identity (whether two variables point to the same object).

## Question 8: What are logical operators in Python?
Logical operators in Python include:
- `and`: Returns `True` if both operands are true.
- `or`: Returns `True` if at least one operand is true.
- `not`: Reverses the truth value.

## Question 9: What is type casting in Python?
Type casting is converting a value from one data type to another, such as `int()`, `float()`, or `str()`.

## Question 10: What is the difference between implicit and explicit type casting?
- Implicit: Done automatically by Python (e.g., integer to float during division).
- Explicit: Done manually by the programmer using functions like `int()`, `float()`, etc.

## Question 11: What is the purpose of conditional statements in Python?
Conditional statements control the flow of execution based on conditions (e.g., `if`, `else`, `elif`).

## Question 12: How does the `elif` statement work?
The `elif` statement checks additional conditions if the previous `if` condition is false. It allows chaining multiple conditions.

## Question 13: What is the difference between `for` and `while` loops?
- `for`: Iterates over a sequence (e.g., list, range).
- `while`: Repeats as long as a condition is true.

## Question 14: Describe a scenario where a `while` loop is more suitable than a `for` loop.
A `while` loop is more suitable when the number of iterations is not predetermined, such as reading user input until a valid value is entered.

## Question 5: Demonstrate mutability of lists and immutability of tuples

```python
my_list = [1, 2, 3]
my_tuple = (1, 2, 3)

# Modify list
my_list[0] = 100
print("Modified list:", my_list)

# Attempt to modify tuple
try:
    my_tuple[0] = 100
except TypeError as e:
    print("Error modifying tuple:", e)
```

## Question 6: Function demonstrating mutable and immutable arguments

```python
def modify_objects(mutable_obj, immutable_obj):
    mutable_obj.append(100)
    immutable_obj += (100,)
    return mutable_obj, immutable_obj

list_obj = [1, 2, 3]
tuple_obj = (1, 2, 3)

modified_list, modified_tuple = modify_objects(list_obj, tuple_obj)
print("Modified list:", modified_list)
print("Modified tuple:", modified_tuple)
```

## Question 13: Reverse a string using while loop

```python
string = "hello"
reversed_string = ""
i = len(string) - 1
while i >= 0:
    reversed_string += string[i]
    i -= 1
print("Reversed string:", reversed_string)
```

## Question 14: Factorial of a number

```python
def factorial(n):
    result = 1
    while n > 1:
        result *= n
        n -= 1
    return result

number = int(input("Enter a number: "))
print("Factorial:", factorial(number))
```
